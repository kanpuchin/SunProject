```python
# download function : retry once again if some file failed

def download_aia_channel(wavelength_angstrom):
    """Search for and download AIA images at one wavelength."""

    wavelength = wavelength_angstrom * u.angstrom

    download_directory = Path(
        f"../data/aia/20110607/{wavelength_angstrom}"
    )

    download_directory.mkdir(parents=True, exist_ok=True)

    print(f"Searching for AIA {wavelength_angstrom} Å images...")

    search_result = Fido.search(
        a.Time(observation_start, observation_end),
        a.Wavelength(wavelength),
        a.Sample(cadence),
        a.jsoc.Series("aia.lev1_euv_12s"),
        a.jsoc.Segment("image"),
        a.jsoc.Notify(jsoc_email),
    )

    print(f"Search results: {len(search_result[0])}")

    download_result = Fido.fetch(
        search_result,
        path=str(download_directory / "{file}"),
        progress=False,
        max_conn=1,
    )

    if download_result.errors:
        print(
            f"Retrying {len(download_result.errors)} "
            f"failed download(s)..."
        )

        download_result = Fido.fetch(
            download_result,
            progress=False,
            max_conn=1,
        )

    image_files = sorted(
        download_directory.glob("*.image_lev1.fits")
    )

    print(f"Downloaded images: {len(image_files)}")
    print(f"Remaining errors: {len(download_result.errors)}")

    return image_files
```

```python
from astropy.io import fits
from astropy.time import Time

target_time = Time("2011-06-07 06:41:00") # largest flare time


def find_closest_observation(files, target_time):
    """Return the observation closest to the target time."""

    observations = []

    for file in files:
        aia_map = sunpy.map.Map(file)

        observation_time = aia_map.date

        time_difference = abs(
            (observation_time - target_time).to_value(u.s)
        )

        observations.append(
            (time_difference, observation_time, file)
        )

    _, observation_time, closest_file = min(
        observations,
        key=lambda item: item[0],
    )

    return closest_file, observation_time
```

```python
def calculate_roi_light_curve(files):
    """Calculate exposure-normalized mean intensity within the ROI."""

    observation_times = []
    mean_intensities = []

    for file in files:
        aia_map = sunpy.map.Map(file)

        bottom_left = SkyCoord(
            600 * u.arcsec,
            -430 * u.arcsec,
            frame=aia_map.coordinate_frame,
        )

        top_right = SkyCoord(
            820 * u.arcsec,
            -260 * u.arcsec,
            frame=aia_map.coordinate_frame,
        )

        roi_map = aia_map.submap(
            bottom_left,
            top_right=top_right,
        )

        exposure_time = aia_map.exposure_time.to_value(u.s)

        normalized_data = roi_map.data / exposure_time

        mean_intensity = np.nanmean(normalized_data)

        observation_times.append(
            aia_map.date.to_datetime()
        )

        mean_intensities.append(mean_intensity)

    return observation_times, np.array(mean_intensities)
```

```python
light_curves = {}

for wavelength in wavelengths:
    observation_times, mean_intensities = (
        calculate_roi_light_curve(
            aia_files[wavelength]
        )
    )

    light_curves[wavelength] = {
        "time": observation_times,
        "intensity": mean_intensities,
    }

    print(
        f"AIA {wavelength} Å: "
        f"{len(mean_intensities)} measurements"
    )
```

```python
import matplotlib.dates as mdates

fig, ax = plt.subplots(figsize=(11, 6))

colors = {
    171: "goldenrod",
    193: "sienna",
    211: "purple",
    304: "crimson",
}

for wavelength in wavelengths:
    times = light_curves[wavelength]["time"]
    intensities = light_curves[wavelength]["intensity"]

    preflare_baseline = np.nanmean(
        intensities[:2]
    )

    relative_intensity = (
        intensities / preflare_baseline
    )

    ax.plot(
        times,
        relative_intensity,
        marker="o",
        linewidth=2,
        color=colors[wavelength],
        label=f"AIA {wavelength} Å",
    )

flare_times = {
    "Flare start": "2011-06-07 06:16:00",
    "GOES peak": "2011-06-07 06:41:00",
    "Flare end": "2011-06-07 06:59:00",
}

for label, time_string in flare_times.items():
    event_time = Time(
        time_string
    ).to_datetime()

    ax.axvline(
        event_time,
        linestyle="--",
        color="gray",
        alpha=0.7,
    )

    ax.text(
        event_time,
        0.98,
        label,
        transform=ax.get_xaxis_transform(),
        rotation=90,
        va="top",
        ha="right",
    )

ax.xaxis.set_major_formatter(
    mdates.DateFormatter("%H:%M")
)

ax.set_xlabel("Observation Time [UTC]")
ax.set_ylabel("Intensity Relative to Pre-flare Baseline")
ax.set_title(
    "AIA Multi-wavelength Flare Light Curves"
)

ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()

fig.savefig(
    "../figures/aia_flare_light_curves.png",
    dpi=200,
    bbox_inches="tight",
)

plt.show()
```

## ROI Light Curves

A smaller photometric region was defined around the flaring active
region to reduce contamination from surrounding structures and
off-limb emission.

For each observation, the mean ROI intensity was normalized by the
exposure time. The resulting light curves were then divided by their
respective pre-flare baseline intensities to compare relative
variations across AIA wavelengths.

## Preliminary Light-curve Results

The AIA channels show distinct temporal responses during the flare.
The 171 and 304 Å intensities rise rapidly and reach their sampled
maxima near 06:25 UTC, whereas the 193 Å intensity reaches its maximum
closer to the GOES soft X-ray peak at 06:41 UTC. The 211 Å channel
shows a more gradual variation.

These results are preliminary. Differences between the light curves
may reflect the temperature response of each channel, changes in
morphology within the selected ROI, image saturation, and the
multi-thermal nature of the AIA passbands. Further calibration and
quality checks are required before drawing a physical conclusion.

```python
import pandas as pd

peak_results = []

for wavelength in wavelengths:
    times = light_curves[wavelength]["time"]
    intensities = light_curves[wavelength]["intensity"]

    baseline = np.nanmean(intensities[:2])
    relative_intensity = intensities / baseline

    peak_index = np.nanargmax(relative_intensity)
    peak_time = times[peak_index]
    peak_intensity = relative_intensity[peak_index]

    peak_results.append(
        {
            "Wavelength (Å)": wavelength,
            "Peak time (UTC)": peak_time.strftime("%H:%M:%S"),
            "Peak relative intensity": peak_intensity,
        }
    )

peak_table = pd.DataFrame(peak_results)

peak_table.style.format(
    {"Peak relative intensity": "{:.2f}"}
)
```

## Results

### Multi-wavelength Morphology

The AIA channels show distinct structures in and around the flaring active region near the GOES soft X-ray peak.

The 171 Å image reveals detailed coronal loops surrounding the bright flare core. The 193 Å image shows enhanced emission from both the central region and the surrounding coronal structures. The 211 Å channel also shows a bright central region together with more diffuse coronal emission. In contrast, the 304 Å image displays relatively compact and elongated bright structures associated with the lower solar atmosphere.

These morphological differences indicate that the four AIA channels sample different plasma components and atmospheric structures during the flare.

### ROI Light Curves

A fixed photometric region of interest was defined over the flaring active region:

* Solar-X: 600–820 arcsec
* Solar-Y: −430 to −260 arcsec

For each observation, the mean intensity within the ROI was calculated. Because the AIA exposure time can vary during a flare, the measured intensity was divided by the exposure time:

$$
I_{\mathrm{ROI}}(t)
=
\operatorname{mean}\left(
\frac{\mathrm{DN}(t)}
{t_{\mathrm{exp}}(t)}
\right)
$$

Each exposure-normalized light curve was then divided by its pre-flare baseline. The baseline was defined as the mean of the first two observations, obtained near 06:05 and 06:10 UTC. The resulting dimensionless light curves describe the intensity of each channel relative to its own pre-flare level.

The channels show clearly different temporal responses. The 171 and 304 Å intensities rise rapidly during the early phase of the event and reach their sampled maxima near 06:25 UTC. The 193 Å intensity increases more gradually and remains enhanced until the GOES soft X-ray peak. The 211 Å channel shows a more moderate and gradual variation.

### Peak-intensity Comparison

The sampled peak times and relative peak intensities are summarized below.

| Wavelength | Sampled peak time (UTC) | Peak intensity relative to baseline |
| ---------: | :---------------------: | ----------------------------------: |
|      171 Å |         06:25:12        |                                1.89 |
|      193 Å |         06:40:10        |                                1.99 |
|      211 Å |         06:35:03        |                                1.63 |
|      304 Å |         06:25:32        |                                5.25 |

The 171 and 304 Å channels reached their sampled maxima near 06:25 UTC. Their peak times differ by only 20 seconds, which is much shorter than the approximately five-minute sampling interval. Therefore, their peak order cannot be resolved with the current dataset.

The 211 Å channel reached its sampled maximum near 06:35 UTC. The 193 Å channel reached its sampled maximum at 06:40:10 UTC, approximately 50 seconds before the GOES soft X-ray peak at 06:41 UTC. This difference is also shorter than the sampling interval, so the sampled 193 Å maximum can be regarded as approximately coincident with the GOES peak.

Among the four channels, 304 Å showed the largest relative enhancement, reaching 5.25 times its pre-flare baseline. The 171, 193, and 211 Å channels reached 1.89, 1.99, and 1.63 times their respective baseline intensities.

## Interpretation and Limitations

The different spatial and temporal behaviors are broadly consistent with the distinct temperature responses and emitting structures sampled by the AIA channels. The 171 Å channel is commonly associated with relatively cool coronal plasma, while the 193 and 211 Å channels are sensitive to hotter coronal emission. The 304 Å channel primarily traces emission from the chromosphere and transition region. During a flare, however, the contribution of different emission lines can change substantially.

Each AIA passband has a broad and sometimes multi-thermal temperature response. Therefore, an AIA channel should not be interpreted as representing plasma at one exact temperature. The observed sequence of sampled maxima does not by itself establish a simple plasma heating or cooling sequence.

The relative peak amplitudes should also not be interpreted as direct measurements of the relative energy emitted in each channel. Each light curve was normalized by its own pre-flare baseline, and the AIA channels have different instrumental responses and absolute intensity scales.

The approximately five-minute sampling interval limits the temporal precision of the analysis. Variations occurring between the selected observations cannot be resolved, and time differences much shorter than five minutes should not be interpreted as significant delays. The observations from different channels are also separated by several to tens of seconds. This is acceptable for comparing the overall evolution but not for measuring sub-minute time delays.

The measured light curves depend on the size and position of the selected ROI. The current ROI includes the flare core and surrounding coronal structures, so the curves represent their combined evolution rather than emission from the flare kernel alone. A smaller ROI could produce a different relative enhancement.

The bright flare core may also contain saturated pixels, although white regions in a displayed image can additionally result from plotting limits. Detector saturation should therefore be checked separately before interpreting the peak intensity quantitatively.

Finally, this preliminary analysis uses Level 1 AIA observations. More precise quantitative and spatial comparisons would require additional pointing correction, image registration, and calibration before the channels are treated as accurately co-aligned observations.
